3. Sales Representatives experience the highest attrition rate among all job roles at approximately 39.8%, meaning nearly 4 out of every 10 employees in this role leave the company.

5. Research Directors and Managers have the lowest attrition rates, suggesting senior-level positions are more stable and less likely to leave the organization.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

lr.fit(X_train, y_train)

print("Logistic Regression Trained Successfully")
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42
)

rf.fit(X_train, y_train)

print("Random Forest Trained Successfully")
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier(
    random_state=42
)

gb.fit(X_train, y_train)

print("Gradient Boosting Trained Successfully")
```

```python
lr_pred = lr.predict(X_test)
rf_pred = rf.predict(X_test)
gb_pred = gb.predict(X_test)

print("Predictions Generated")
```

```python
from sklearn.metrics import (
    classification_report,
    roc_auc_score
)
```

```python
print("LOGISTIC REGRESSION")
print("="*50)

print(
    classification_report(
        y_test,
        lr_pred
    )
)
```

```python
print("RANDOM FOREST")
print("="*50)

print(
    classification_report(
        y_test,
        rf_pred
    )
)
```

```python
print("GRADIENT BOOSTING")
print("="*50)

print(
    classification_report(
        y_test,
        gb_pred
    )
)
```

```python
lr_prob = lr.predict_proba(X_test)[:,1]

rf_prob = rf.predict_proba(X_test)[:,1]

gb_prob = gb.predict_proba(X_test)[:,1]
```

```python
print(
    "Logistic Regression ROC-AUC:",
    roc_auc_score(
        y_test,
        lr_prob
    )
)

print(
    "Random Forest ROC-AUC:",
    roc_auc_score(
        y_test,
        rf_prob
    )
)

print(
    "Gradient Boosting ROC-AUC:",
    roc_auc_score(
        y_test,
        gb_prob
    )
)
```

```python
results = {
    
    "Model":[
        "Logistic Regression",
        "Random Forest",
        "Gradient Boosting"
    ],
    
    "ROC_AUC":[
        roc_auc_score(y_test, lr_prob),
        roc_auc_score(y_test, rf_prob),
        roc_auc_score(y_test, gb_prob)
    ]
}

comparison = pd.DataFrame(results)

comparison.sort_values(
    by="ROC_AUC",
    ascending=False
)
```

```python
from sklearn.metrics import roc_curve
```

```python
fpr_lr, tpr_lr, _ = roc_curve(
    y_test,
    lr_prob
)

fpr_rf, tpr_rf, _ = roc_curve(
    y_test,
    rf_prob
)

fpr_gb, tpr_gb, _ = roc_curve(
    y_test,
    gb_prob
)

plt.figure(figsize=(8,6))

plt.plot(
    fpr_lr,
    tpr_lr,
    label="Logistic Regression"
)

plt.plot(
    fpr_rf,
    tpr_rf,
    label="Random Forest"
)

plt.plot(
    fpr_gb,
    tpr_gb,
    label="Gradient Boosting"
)

plt.plot(
    [0,1],
    [0,1],
    "--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve Comparison")

plt.legend()

plt.tight_layout()

plt.savefig(
    "charts/roc_curve.png"
)

plt.show()
```

## Final Business Insights

1. Employees working as Laboratory Technicians show the highest likelihood of leaving the organization.

2. Employees who frequently work overtime are significantly more likely to leave than employees who do not work overtime.

3. Employees who travel frequently for business demonstrate higher attrition rates than employees with less travel.

4. Sales-related roles and the Sales department consistently show elevated employee turnover compared to other parts of the company.

5. Employees who have not received promotions for long periods are more likely to leave, suggesting career growth opportunities influence retention.

6. Lower-income employees exhibit higher attrition rates, although salary alone does not fully explain employee exits.

7. Employees with poor work-life balance are considerably more likely to leave the organization.

## HR Recommendations

### Recommendation 1

Implement targeted retention programs for Laboratory Technicians and Sales Representatives. Regular career discussions, mentorship programs, and development opportunities should be prioritized for these high-risk employee groups.

### Recommendation 2

Reduce excessive overtime and improve work-life balance policies. Employees working frequent overtime should receive workload reviews and manager check-ins to reduce burnout and improve retention.

## Model Limitation

This model is trained using historical employee data and cannot capture all factors that influence employee decisions. Personal circumstances, organizational culture changes, external job market conditions, and economic factors are not included in the dataset. Therefore, predictions should be used as a decision-support tool rather than a replacement for human judgment.
